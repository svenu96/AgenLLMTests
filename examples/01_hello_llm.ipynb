## 1. 🤖 Initialize Your First LLM

We'll start with a small, manageable model that you can run locally.

```python
# Initialize LLM with a small model for learning
llm = LLMBasics(model_name="distilbert-base-uncased")

# Check if the model loaded successfully
if llm.model:
    print("✅ Great! Your LLM is ready to explore.")
    print(f"📊 Model: {llm.model_name}")
    print(f"📝 Vocabulary size: {llm.tokenizer.vocab_size:,} tokens")
else:
    print("❌ Model loading failed. Don't worry, we'll show you the concepts anyway!")
```

## 2. 🔤 Understanding Tokenization

Tokenization is how computers break down text into pieces they can understand.

```python
# Let's explore how text gets tokenized
sample_texts = [
    "Hello, world!",
    "Machine learning is fascinating!",
    "The quick brown fox jumps over the lazy dog.",
    "Artificial intelligence will change everything."
]

print("🔍 Tokenization Examples:")
print("=" * 50)

for text in sample_texts:
    print(f"\n📝 Text: '{text}'")
    
    if llm.tokenizer:
        result = llm.tokenization_demo(text)
        print(f"🔢 Tokens: {result['tokens'][:10]}{'...' if len(result['tokens']) > 10 else ''}")
        print(f"📊 Token count: {len(result['tokens'])}")
    else:
        # Show concept even without model
        words = text.split()
        print(f"🔢 Words: {words}")
        print(f"📊 Word count: {len(words)}")
        print("💡 Note: Real tokenization is more sophisticated than word splitting!")
```

## 3. 🎯 Attention Mechanisms - The Heart of Transformers

Attention helps models focus on relevant parts of the input.

```python
# Demonstrate attention concepts
concepts = TransformerConcepts()

print("🧠 Understanding Attention:")
attention_demo = concepts.attention_mechanism_demo()

print("\n💡 Key Insights:")
print("- Queries ask 'what am I looking for?'")
print("- Keys answer 'what do I contain?'")
print("- Values provide 'what information do I have?'")
print("- Attention weights determine focus")
```

```python
# Visualize attention patterns if model is available
if llm.model:
    print("🎨 Visualizing Attention Patterns:")
    
    sample_text = "The cat sat on the mat and looked around."
    print(f"Text: '{sample_text}'")
    
    try:
        attention_result = llm.attention_visualization(sample_text, layer=0)
        print("✅ Attention visualization complete!")
        print("💡 Darker colors = stronger attention connections")
    except Exception as e:
        print(f"⚠️ Visualization not available: {e}")
        print("💡 This would show how words attend to each other")
else:
    print("🎨 Attention Visualization (Conceptual):")
    print("In a real attention visualization, you'd see:")
    print("- A heatmap showing word-to-word attention")
    print("- Stronger connections between related words")
    print("- How the model decides what's important")
```

## 4. 🎨 Word Embeddings - Capturing Meaning

Word embeddings convert words into numerical vectors that capture semantic meaning.

```python
# Explore word embeddings
if llm.model:
    print("🎨 Exploring Word Embeddings:")
    
    # Words with different semantic relationships
    word_groups = {
        "Emotions": ["happy", "sad", "angry", "excited", "calm"],
        "Animals": ["cat", "dog", "bird", "fish", "horse"],
        "Technology": ["computer", "phone", "internet", "software", "data"]
    }
    
    for category, words in word_groups.items():
        print(f"\n📊 {category} Words:")
        print(f"Words: {', '.join(words)}")
        
        try:
            embedding_result = llm.embedding_exploration(words)
            print("✅ Embedding analysis complete!")
            print("💡 Similar words should have higher similarity scores")
        except Exception as e:
            print(f"⚠️ Embedding analysis not available: {e}")
            break
        
        if category == "Emotions":  # Only show one visualization to save time
            break
else:
    print("🎨 Word Embeddings (Conceptual):")
    print("Word embeddings would show:")
    print("- Similar words clustered together")
    print("- Relationships captured in vector space")
    print("- Semantic meaning preserved numerically")
```

## 5. 📍 Positional Encoding - Understanding Sequence

Since attention doesn't inherently understand word order, we need positional encoding.

```python
# Demonstrate positional encoding
print("📍 Understanding Positional Encoding:")

# Create positional encoding visualization
pos_encoding = concepts.positional_encoding_demo(seq_len=20, d_model=16)

print("\n💡 Key Points:")
print("- Each position gets a unique encoding")
print("- Sinusoidal patterns help models understand relative positions")
print("- Added to word embeddings to preserve sequence information")
print("- Essential for understanding sentence structure")
```

## 6. 💬 Basic Prompt Engineering

Learn how to craft effective prompts for better results.

```python
print("💬 Prompt Engineering Basics:")
print("=" * 40)

# Examples of prompt engineering principles
prompt_examples = {
    "❌ Vague": "Tell me about AI",
    "✅ Specific": "Explain the main differences between supervised and unsupervised machine learning in 3 bullet points",
    
    "❌ No Context": "How do I fix this?",
    "✅ With Context": "I'm getting a 'module not found' error in Python when importing pandas. How do I fix this?",
    
    "❌ No Format": "List some programming languages",
    "✅ Clear Format": "List 5 popular programming languages with their primary use cases in this format: Language - Use Case"
}

for label, prompt in prompt_examples.items():
    print(f"\n{label}:")
    print(f"'{prompt}'")

print("\n🎯 Prompt Engineering Tips:")
print("1. Be specific about what you want")
print("2. Provide context and background")
print("3. Specify the desired format")
print("4. Give examples when helpful")
print("5. Break complex tasks into steps")
```
